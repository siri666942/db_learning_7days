```python
# k
def f(n):
    if n==0:
        return 10
    else:
        return 60+1.2*f(n-1)
f(3)
```

```python
def g(n):
    res=10*(1.2**n)
    for i in range(n):
        res+=60*(1.2**i)
    return res
g(3)
```

```python
def get_month_return(n):
    ''' n belike :0.05,0.1,0.2;'''
    # 1+n=(1+x)^12
    return (1+n)**(1/12)-1

get_month_return(0.1)
```
