| API | PostgreSQL |
| --- | --- |
|tokenInfo/tokenAbbr|token|
| hash | hash |
| block | block_number |
| from | from_address |
| to | to_address |
| amount | value |
| block_timestamp | created_at |
| 执行成功与否,contract_ret | contract_ret |
| contractType | contract_type |

```python
from datetime import  datetime,timezone
def handle_data(data:dict):
    txs=data['data']
    token=data['tokenInfo']['tokenAbbr']
    outputs=[]
    for raw_tx in txs:
        tx={
            'hash':raw_tx['hash'],
            'token':token,
            'block_number':raw_tx['block'],
            'from_address':raw_tx['from'],
            'to_address':raw_tx['to'],
            'value':int(raw_tx['amount']),
            'created_at':datetime.fromtimestamp(
                raw_tx['block_timestamp']/1000,
                tz=timezone.utc
                
                ),
            'contract_ret':0 if raw_tx['contract_ret']=='SUCCESS' else 1,
            'contractType':raw_tx['contractType'],
        }
        outputs.append(tx)
    return outputs
        
```

```python
clean_txs=handle_data(data)
```

```python
import psycopg

conn = psycopg.connect(
    host="localhost",
    port=5432,
    dbname="tron_binance_txs",
    user="siri",
    password=os.environ["POSTGRES_PASSWORD"],
)
cur = conn.cursor()
```

```python
sql='''create table tron_transactions(
    hash CHAR(64) PRIMARY KEY,
    token VARCHAR(16) NOT NULL,
    block_number  BIGINT NOT NULL,
    from_address VARCHAR(64) NOT NULL,
    to_address VARCHAR(64) NOT NULL,
    value NUMERIC(78,0) NOT NULL,
    created_at TIMESTAMPTZ NOT NULL,
    contract_ret SMALLINT NOT NULL,
    contract_type INTEGER NOT NULL,
    CHECK (contract_ret in (0,1))
);'''

cur.execute(sql)
conn.commit()
```

```python
cur.execute("comment on column tron_transactions.contract_ret is '交易结果，0 表示成功，1 表示失败；';")
cur.execute('''CREATE INDEX idx_tron_transactions_created_at
ON tron_transactions(created_at);''')
cur.execute('''create INDEX idx_tron_transactions_token_created_at
ON tron_transactions(token,created_at);''')
conn.commit()
```

```python
# view :token=trx,把单位换算成 TRX。并只保留成功的交易。
sql="""
create view successful_txs_by_trx AS
select 
    hash,block_number,from_address,to_address,value/ 1000000.0 AS amount_by_trx,created_at,contract_type
from tron_transactions
where contract_ret=0 and token='trx';
"""

cur.execute(sql)
conn.commit()
```
